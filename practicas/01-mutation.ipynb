```python
import unittest

def pre_condicion(condicion: bool):
    if not condicion:
        raise Exception("No satisface pre-condición")
```

## Ejercicio 1

```python
def foo(a: list[int], b: list[int]) -> int:
    pre_condicion(hasattr(a, '__len__') and # Evita TypeError
                  len(a) <= len(b)) # Evita IndexError

    r: int = 0
    i: int = 0
    while i < len(a):
        r += a[i] * b[i]
        i += 1
    return r
```

```python
foo(9, [4, 5, 6])
```

## Ejercicio 2

```python
def foo_2(x: int, y: int) -> int:
    z: int = 0
    if x <= y:
        z = x
    else:
        z = y
    return z
```

```python
class TestFoo2(unittest.TestCase):
    def test_1(self):
        rv = foo_2(1, 0)
        self.assertEqual(rv, 0)

    def test_2(self):
        rv = foo_2(0, 1)
        self.assertEqual(rv, 0)

unittest.TextTestRunner().run(unittest.TestLoader().loadTestsFromTestCase(TestFoo2))
```

## Ejercicio 3

```python
def foo_2_mut(x: int, y: int) -> int:
    z: int = 0
    if x != y:  # Mutación
        z = x
    else:
        z = y
    return z
```

```python
class TestFoo2(unittest.TestCase):
    def test_1(self):
        self.assertEqual(foo_2_mut(0, 1), 0)

    def test_2(self):
        self.assertEqual(foo_2_mut(0, 0), 0)

    def test_3(self):
        self.assertEqual(foo_2_mut(1, 0), 0) # Se espera 0 en foo_2, revuelve 1 en el mutante

unittest.TextTestRunner().run(unittest.TestLoader().loadTestsFromTestCase(TestFoo2))
```

## Ejercicio 7

```python
def testme_2(k: int, j: int) -> int:
    i: int = 0
    r: int = 0
    while i < 2:  # C1
        if k == j:  # C2
            r = r + 1
        i = i + 1
    return r
```

### a)

Mutante de operador absoluto (modifica una expresión artimética E reemplazándola por abs(E), −abs(E), o la constante 0):

```python
def testme_2_mut(k: int, j: int) -> int:
    i: int = 0
    r: int = 0
    while i < 2:  # C1
        if k == j:  # C2
            r = 0 # Mutación
        i = i + 1
    return r
```

```python
funcion_a_testear = testme_2

class TestTestMe2(unittest.TestCase):
    def __init__(self, testCaseName='runTest'):
        super().__init__(testCaseName)

    def test_case_1(self):
        self.assertEqual(funcion_a_testear(5, 5), 2)

    def test_case_2(self):
        self.assertEqual(funcion_a_testear(3, 3), 2)

    def test_case_3(self):
        self.assertEqual(funcion_a_testear(1, 0), 0)
```

```python
unittest.TextTestRunner().run(unittest.TestLoader().loadTestsFromTestCase(TestTestMe2))
```

Este mutante reemplaza la operación `r+1` por `0`, haciendo que retorne siempre `0`.

```python
funcion_a_testear = testme_2_mut
unittest.TextTestRunner().run(unittest.TestLoader().loadTestsFromTestCase(TestTestMe2))
```

### b)

Mutante equivalente:

```python
def testme_2_mut_equivalente(k: int, j: int) -> int:
    i: int = 0
    r: int = 0
    while i < 2:  # C1
        if k == j:  # C2
            r = r + 1
        i = abs(i + 1) # mutación
    return r
```

Este mutante es equivalente a la función original ya que realiza un `abs()` sobre un operando que siempre será positivo.

```python
funcion_a_testear = testme_2_mut_equivalente
unittest.TextTestRunner().run(unittest.TestLoader().loadTestsFromTestCase(TestTestMe2))
```
